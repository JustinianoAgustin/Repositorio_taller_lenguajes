```python
grafico = tabla.sort_values(by='Rank', ascending=True).head(10)
```

```python
grafico1 = grafico.sort_values(by='TIOBE_Share_Pct', ascending=True)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
grafico1.plot.barh(
    x='Language',
    y='TIOBE_Share_Pct',
    ax=ax,
    color='#1f77b4',
    edgecolor='black',
    legend=False,
)
ax.set_title('Los 10 Lenguajes de Programación Más Demandados (Cuota TIOBE)')
```
